```python
# Step 1: Import libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
# Step 2: Load dataset

df = pd.read_excel("dummy_wheat_yield_380_fields.xlsx")

print(df.head())
print("Dataset Shape:", df.shape)
```

```python
# Step 3: Select input features (X) and target (y)

X = df[
    [
        "NDVI",
        "NDRE",
        "Rainfall_mm_seasonal",
        "Mean_Temperature_C",
        "Mean_Solar_Radiation_MJ_m2_day"
    ]
]

y = df["Yield_t_ha"]
```

```python
# Step 4: Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Step 5: Define 3 regression models

models = {
    "Linear Regression": make_pipeline(
        StandardScaler(),
        LinearRegression()
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}
```

```python
# Step 6: Train and evaluate models

results = []

for name, model in models.items():

    # Training
    model.fit(X_train, y_train)

    # Prediction
    predictions = model.predict(X_test)

    # Evaluation
    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })
```

```python
# Step 7: Compare results

results_df = pd.DataFrame(results)

print("\nModel Comparison:")
print(results_df.sort_values(by="R2 Score", ascending=False))
```

```python

import joblib

# Best model select karo
best_model = models["Gradient Boosting"]

# Save trained model
joblib.dump(best_model, "wheat_yield_model.pkl")

print("Model saved successfully!")
```
